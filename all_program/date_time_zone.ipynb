```python
import datetime 
my_date = datetime.datetime.now()

# day add
day_date= my_date + datetime.timedelta(days=5)

# month add
month = my_date.month + 2 # ahiya month add kari shakiye
year = my_date.year
if month>12:
    month=1
    year+=1
month_date= my_date.replace(year = year, month = month)

# year add
year_date = month_date.replace(year = my_date.year + 1)

print("now date:",my_date)
print("add 5 day:",day_date)
print("add 2 month",month_date)
print("add 1 year:",year_date)
```

```python
import datetime
my_time = datetime.time(8,30,00)

time_replace = my_time.replace(hour=1,minute=10,second=20) # current time ne replace kare

dt=datetime.datetime.combine(datetime.date.today(), my_time)
time_add = dt + datetime.timedelta(hours=1,minutes=10,seconds=20)

print("set time:",my_time)
print("replace time:",time_replace)
print("added time:",time_add)
```

```python
import datetime
from pytz import timezone
from dateutil.relativedelta import relativedelta

utc_time = datetime.datetime.now(datetime.timezone.utc)
india_time = utc_time.astimezone(timezone("Asia/Kolkata"))

add_day = india_time + datetime.timedelta(days=1)

add_month = india_time + relativedelta(months=2)

add_year = india_time + relativedelta(years=1)

# dt = datetime.datetime.combine(datetime.date.today(),india_time.time())
# add_hour= dt + datetime.timedelta(hours=2)
# add_minute = dt + datetime.timedelta(minutes=20)
# add_second = dt + datetime.timedelta(seconds=30)

add_hour = india_time + datetime.timedelta(hours=2)
add_minute = india_time + datetime.timedelta(minutes=20)
add_second = india_time + datetime.timedelta(seconds=30)
print("UTC Time:",utc_time)
print("INDIA Time:",india_time)
print("Added 1 Day:",add_day)
print("Added 2 Month:",add_month)
print("Added 1 Year:",add_year)
print("Added 2 Hour:",add_hour)
print("Added 20 Minute:",add_minute)
print("Added 30 Second:",add_second)
```

```python
import datetime
from dateutil.relativedelta import relativedelta

today= datetime.datetime.now()

add_day = today + datetime.timedelta(days = 2)
remove_day = today + datetime.timedelta(days = -2)

add_month = today + relativedelta(months= 2)
remove_month = today + relativedelta(months= -2)

add_year = today + relativedelta(years= 2)
remove_year = today + relativedelta(years= -2)

add_hour = today + datetime.timedelta(hours = 2)
remove_hour = today + datetime.timedelta(hours = -2)

add_minute = today + datetime.timedelta(minutes = 2)
remove_minute = today + datetime.timedelta(minutes = -2)
add_second = today + datetime.timedelta(seconds = 2)
remove_second = today + datetime.timedelta(seconds = -2)

print("MAIN DATE",today)
print("add day + 2:",add_day)
print("remove day - 2:",remove_day)
print("add month + 2:",add_month)
print("remove month - 2:",remove_month)
print("add year + 2:",add_year)
print("remove year - 2:",remove_year)
print("add hour + 2:",add_hour)
print("remove hour - 2:",remove_hour)
print("add minute + 2:",add_minute)
print("remove minute - 2:",remove_minute)
print("add second + 2:",add_second)
print("remove second - 2:",remove_second)
```
