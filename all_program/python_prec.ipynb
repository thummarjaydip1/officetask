```python
class Parent():
    def __init__(self):
        print("parent")

    def displayparent(self):
        print("Parent class display method")

class Child(Parent):
    def __init__(self):
        print("child")

    def displaychild(self):
        print("Child class display method")

obj = Child()
obj.displaychild()
obj.displayparent()
```

```python
# Method overidding
class Parent():
    def display(self):
        print("parent display")
    
class Child(Parent):
    def display(self):
        print("child display")

obj = Child()
obj.display()
```

```python
# Methodd overloading
class Calculate():
    def addition(self, a=0, b=0, c=0):
        return a + b + c 
obj = Calculate()
print(obj.addition())
print(obj.addition(10))
print(obj.addition(10,20))
print(obj.addition(10,20,30))
print(obj.addition(21,28,51))
```

```python
class Myclass():
    def __init__(self):
        print("__init__ method")
    
    def add(self, a, b):
        print("First number:", a)
        print("Second number:", b)
        print("Addition:", a + b)

    def sub(self, a, b):
        print("First number:", a)
        print("Second number:", b)
        print("Substraction:", a - b)

    def mul(self, a, b):
        print("First number:", a)
        print("Second number:", b)
        print("Multiplication:", a * b)

    def div(self, a, b):
        print("First number:", a)
        print("Second number:", b)
        print("Division:", a/b)

c1 = Myclass()
print("--------------------")
c1.add(20, 10)
print("--------------------")
c1.sub(20, 10)
print("--------------------")
c1.mul(20, 10)
print("--------------------")
c1.div(20, 10)
```

```python
class Addition():
    
    def addition(self, a, b):
        return a + b 
    
class Substraction(Addition):
    
    def substraction(self, a, b):
        return a - b
    
class Multiplication(Substraction):
    
    def multiplication(self, a, b):
        return a * b
    
class Division(Multiplication):

    def division(self, a, b):
        return a / b
    
obj = Division()

print( "Addition is :", obj.addition(200, 100) )
print( "Substraction is :", obj.substraction(200, 100) )
print( "Multiplication is :", obj.multiplication(200, 100) )
print( "Division is :", obj.division(200,100) )
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n-1)
    
print(factorial(5))
```

```python
import datetime
from pytz import timezone

def timefun(name):
    
    utc_time = datetime.datetime.now(datetime.timezone.utc)
    country_time = utc_time.astimezone(timezone(name))

    print("UTC Time :", utc_time)
    print(f"{name} :", country_time)

timefun("asia/kolkata")
print("----------------------------------------------------")
timefun("America/new_york")
```
