```python
class Calculator():
    def add(self, num1, num2):
        return num2 + num1
    
    def sub(self, num1, num2):
        return num2 - num1
    
    def mul(self, num1, num2):
        return num2 * num1
    
    def div(self, num1, num2):
        return num2/num1
    
c1 = Calculator()
print("addition is:", c1.add(10,20))
print("subscription is:", c1.sub(10,20))
print("multiplication is:", c1.mul(10,20))
print("division is:", c1.div(10,20))
```

```python
class Person():
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def display(self):
        print("Name is: " + self.name)
        print("Age is: " + self.age)

class Myclass(Person):
    def __init__(self, name, age):
        self.name = name
        self.age = age

c1 = Myclass("jaydip", str(22))
c1.display()
```

```python
# Polymorphism Example
class Student1():
    def __init__(self, name, city):
        self.name = name
        self.city = city
    
    def display(self):
        print("name is:", self.name)
        print("city is:", self.city)
    
    def same_method(self):
        print("Student 1 detail")
    
class Student2():
    def __init__(self, name, city):
        self.name = name
        self.city = city
    
    def display(self):
        print("name is:", self.name)
        print("city is:", self.city)

    def same_method(self):
        print("Student 2 detail")

class Student3():
    def __init__(self, name, city):
        self.name = name
        self.city = city
    
    def display(self):
        print("name is:", self.name)
        print("city is:", self.city)
    
    def same_method(self):
        print("Student 3 detail")

class Student4():
    def __init__(self, name, city):
        self.name = name
        self.city = city

    def display(self):
        print("name is:", self.name)
        print("city is:", self.city)

    def same_method(self):
        print("Student 4 detail")

stud1 = Student1("jay", "amreli")
stud2 = Student2("dip", "surat")
stud3 = Student3("jaydip", "ahemdabad")
stud4 = Student4("jayesh", "pune")

for i in (stud1, stud2, stud3, stud4):
    i.same_method()
    i.display()
    print()
```

```python
# Encapsulation 
class Myclass():
    def __init__(self, name, age):
        self.name = name
        self.__age = age
    
    def get_name(self):
        return self.name
    
    def get_age(self):
        return self.__age

c1 = Myclass("jaydip",21)
print(c1.get_name())
print(c1.get_age())
```

```python
# Inner Class
class Outer():
    def __init__(self):
        self.name = "Outer Class"
    
    class Inner():
        def __init__(self):
            self.name = "Inner Class"
        
        def display(self):
            print("display method call")

outer = Outer()
print(outer.name)
outer.Inner.display(outer)

print()

inner = outer.Inner()
print(inner.name)
inner.display()
```

```python
class Outer():
    def __init__(self, name):
        self.name = f"This is {name} class"

    def display(self):
        return "Outer class display method"

    class Inner():
        def __init__(self, name):
            self.name = f"This is {name} class"
        
        def display(self):
            return "Inner class display method"

print("=======Outer Class========")
outer = Outer("Outer")
print(outer.name)
print(outer.display())

print()

print("=======Inner Class========")
inner = outer.Inner("Inner")
print(inner.name)
print(inner.display())
```
