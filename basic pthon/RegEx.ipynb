## Python RegEx

```python
import re

txt = "Hello user welcome my profile"
x = re.search("^Hello.*profile$", txt)
y = re.search("^Hello.+profile$", txt)
print(x)
print(y)
```

```python
import re

txt = "Hello user welcome my profile"
x = re.search("^Hello.................profile$", txt)
print(x)
```

```python
import re
txt = "hello user welcome"
x = re.findall('user|users', txt)
print(x)
if x:
    print("yes, word are available in string")
else:
    print("no, word are not available in string")
```

```python
import re

txt = "hello user welcome to my profile"
x = re.findall("\Ahello", txt)
print(x)
if x:
    print("yes")
else:
    print("no")
```

```python
import re 

txt = "hello user welcome to my profile 21"
x = re.findall("\d",txt)                    # number print only number find
print(x)
if x: 
    print("yes")
else:
    print("no")
```

```python
import re 

txt = "hello user welcome to my profile 21"
x = re.findall("\D",txt)         # number not print only string find
print(x)
```

```python
import re

txt = "hello user welcome to my profile"
x = re.findall("\s", txt)                   # return only space character
print(x)
```

```python
import re

txt = "hello user welcome to my profile"
x = re.findall("\S", txt)               # return only string not white space
print(x)
```

```python
import re 

txt = "hello user welcome to my 21 number profile"
x = re.findall("\w", txt)                       # return all string like digit and charcter
print(x)
```

```python
import re
txt = "hello user welcome to my 21 & number profile"
x = re.findall("\W", txt)                       # return only white space or special character not digit and alphabet
print(x)
```
