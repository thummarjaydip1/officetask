```python
fruit = ["apple", "banana", "mango", "kiwi", "lemon", "pineapple"]

print("-------for loop-------")
for i in fruit:
    print(i)

print("-------for loop with range-------")
for i in range(len(fruit)):
    print(fruit[i])

print("-------while loop-------")
a = 0 
while a < len(fruit):
    print(fruit[a])
    a+=1
```

```python
import array 
a = [10,20,30,40,50,60,70,20]
b = array.array("i", a)
print(b)                    # print array

c = b[2:5]
print(c)                    # slicing array

d = b.index(50)
print(d)                    # searching array(index() method)

b[3] = 100
print(b)                    # update array

e = b.count(20)
print(e)                    # counting array element(count() method)

b.remove(30)
print(b)                    # removing array element(remove() method)

b.pop()
print(b)                    # removing array last element(pop() method) and use specific index remove item

b.insert(2, 30)
print(b)                    # array in insert element in specific position(insert() method)

b.reverse()
print(b)                    # array element descending order(reverse() method)

b.clear()
print(b)                    # all array element clear(clear() method)
```
