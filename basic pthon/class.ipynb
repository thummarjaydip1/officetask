##### Polymorphism in class

```python
# compile time polymorphism
class Myclass:
    def multiply(self, a=1, b=1, *args):
        result = a * b
        for i in args:
            result = result * i
        return result

obj = Myclass()
print(obj.multiply())
print(obj.multiply(12))
print(obj.multiply(2,5))
print(obj.multiply(1,2,3,4,5))
```

```python
# polymorphism with inheritance
class Animal:
    def __init__(self, name):
        self.name = name

    def sound(self):
        print("sound")

class Cat(Animal):
    pass

class Dog(Animal):
    def sound(self):
        print("brak")

class Horse(Animal):
    def sound(self):
        print("hihihi...")

c = Cat("kitty")
d = Dog("sheru")
h = Horse("king")

for i in (c, d, h):
    print(i.name)
    i.sound()
    print()
```

```python
# polymorphism with inheritance
class Calculate:
    def __init__(self, a, b):
        self.a = a
        self.b = b

class Addition(Calculate):
    def cal(self):
        return f"Addition is: {self.a + self.b}"

class Substraction(Calculate):
    def cal(self):
        return f"Substraction is: {self.a - self.b}"

class Multiplication(Calculate):
    def cal(self):
        return f"Multiplication is: {self.a * self.b}"

class Division(Calculate):
    def cal(self):
        return f"Division is: {self.a / self.b}"

add = Addition(20, 10)
sub = Substraction(20, 10)
mul = Multiplication(20, 10)
div = Division(20, 10)

all_obj = (add, sub, mul, div)

for i in all_obj:
    print(i.cal())
```

##### Inheritance in Python

```python
class Myclass:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        return f"name is: {self.name} and age is: {self.age}"

class Person(Myclass):
    pass

p1 = Person("jay", 21)
print(p1.display())
```

```python
# super() method
class Myclass:
    def __init__(self, name):
        self.name = name

    def disp(self):
        print("name is:", self.name)

class Person(Myclass):
    def __init__(self, name, age):
        super().__init__(name)
        self.age = age

    def display(self):
        print("name is:", self.name, "age is:", self.age)

p1 = Person("jay", 21)
p1.disp()
p1.display()
```

```python
# multiple inheritance
class Father:
    def father(self):
        print("hello father")

class Mother:
    def mother(self):
        print("hello other")

class Child(Father, Mother):
    def child(self):
        print("hello child")

c1 = Child()
c1.child()
c1.mother()
c1.father()
```

```python
class A:
    def show(self):
        print("class A")

class B(A):
    def show(self):
        print("class B")
        super().show()

class C(A):
    def show(self):
        print("class C")
        super().show()

class D(C, B):
    def show(self):
        print("class D")
        super().show()

obj = D()
obj.show()
```

```python
# protected datamember
class Myclass:
    def __init__(self, name):
        self._name = name               # protected member

    def disp(self):
        print("inside within class:", self._name)

class Person(Myclass):
    def display(self):
        print("inside subclass:", self._name)

p1 = Person("jay")
p1.disp()
p1.display()
```

```python
# public protected and private method
class Bankaccount:
    def __init__(self):
        self.balance = 2000

    def _show_balance(self):
        print("account balance is:", self.balance)

    def __add_balance(self, amount):
        self.balance = self.balance + amount

    def __remove_balance(self, amount):
        self.balance = self.balance - amount

    def credit_debit(self, arg, amount):

        if arg == "credit":
            if amount > 0:
                self.__add_balance(amount)
                print("credit amount is:", amount)
                self._show_balance()
            else:
                print("enter amount positive")

        elif arg == "debit":
            if amount > 0:
                if self.balance >= amount:
                    self.__remove_balance(amount)
                    print("debit amount is:", amount)
                    self._show_balance()
                else:
                    print("Balance bellow to amount please check balance")
            else:
                print("enter amount positive")

        else:
            print("please enter ('credit', 'debit')")

obj = Bankaccount()
obj._show_balance()
print()

obj.credit_debit("credit", 500)
print()

obj.credit_debit("debit", 1000)
print()
```
