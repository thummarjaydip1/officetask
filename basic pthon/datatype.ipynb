## Datatype

```python
a = 12
b = -12
print(type(a))
print(type(b))
```

```python
a = 12.0
b = -12.43
print(type(a))
print(type(b))
```

```python
a = None
print(type(a))
```

```python
a = 5j
b = 5 + 32j
print(type(a))
print(type(b))
```

```python
a = True
b = False
print(type(a))
print(type(b))
```

```python
a = "hello how are you?"
b = "123"
c = "423.23"
print(type(a))
print(type(b))
print(type(c))
```

```python
a = range(1,5,1)
print(type(a))
```

```python
value = bytearray([74, 97, 121, 100, 105, 112])  
print(value)
print(type(value))
```

```python
value = bytes([74, 97, 121, 100, 105, 112])
print(value)
print(type(value))
```

```python
data = bytearray(b'hello')
view = memoryview(data)
print(view)
print(type(view))
```

```python
a = {
    "name" : "jay",
    "age" : 21,
    "city" : "surat"
}
print(a)
print(type(a))
```

```python
a = 12
print(a)
print(type(a))

b = float(a)
print(b)
print(type(b))

c = str(a)
print(c)
print(type(c))

d = 12.435
e = int(d)
print(e)
print(type(e))
```
