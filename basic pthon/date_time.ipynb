```python
import datetime
from pytz import timezone
from dateutil.relativedelta import relativedelta

utc_time = datetime.datetime.now(datetime.timezone.utc)
india_time = utc_time.astimezone(timezone("asia/kolkata"))

add_day = india_time + datetime.timedelta(days=5)
add_month = india_time + relativedelta(months=4)
add_year = india_time + relativedelta(years=4)

add_hour = india_time + datetime.timedelta(hours=3)
add_minute = india_time + datetime.timedelta(minutes=30)
add_second = india_time + datetime.timedelta(seconds=30)

print(utc_time)
print(india_time)
print(add_day)
print(add_month)
print(add_year)
print(add_hour)
print(add_minute)
print(add_second)
```
