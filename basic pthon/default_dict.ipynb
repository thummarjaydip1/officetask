## Defaultdict in Python

```python
from collections import defaultdict
d = defaultdict(list)
d["fruits"] = "apple"
d["vagitables"] = "tomato"
print(d)
print(d["juice"])
```

```python
from collections import defaultdict
d = defaultdict(list)

for i in range(5):
    d[i].append(i)

print("dictionary with value as list:")
print(d)
```

```python
from collections import defaultdict
d = defaultdict(int)

nums = [1,2,3,4,2,3,4,1,2]

for i in nums:
    d[i] += 1

print(d)
print(d[5])
```

```python
from collections import defaultdict
d = defaultdict(str)

d["fruits"] = "apple"
print(d)
print(d["vahical"])
```

```python
from collections import defaultdict

d = defaultdict(list)
word = ["apple","ant","banana","bat","carrot","cat"]

for i in word:
    d[i[0]].append(i)

print(d)
```
