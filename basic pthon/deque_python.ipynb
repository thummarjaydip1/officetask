## Deque Python

```python
from collections import deque
d = deque(["jay","dip","man","ram","raj"])
print(d)
```

```python
from collections import deque
d = deque([10,20,30,40,50,60])
print(d[0])
print(d[3])
print(d[-1])
```

```python
from collections import deque
d = deque([10,20,30,40])
d.append(50)
print(d)
```

```python
from collections import deque
d = deque([10,20,30,40])
d.appendleft(50)
print(d)
```

```python
from collections import deque
d = deque([10,20,30,40])
d.extend([50,60,70])
print(d)
```

```python
from collections import deque
d = deque([10,20,30,40])
d.extendleft([50,60,70])
print(d)
```

```python
from collections import deque
d = deque([10,20,30,40])
d.remove(20)
print(d)
```

```python
from collections import deque
d = deque([10,20,30,40,50])
d.pop()
print(d)
```

```python
from collections import deque
d = deque([10,20,30,40,50])
d.popleft()
print(d)
```

```python
from collections import deque
d = deque([11,22,33,44,55])
d.clear()
print(d)
```

```python
from collections import deque
d = deque([10,20,30,40,50,60])
print(len(d))
```

```python
from collections import deque
d = deque([10,20,30,20,40,50,20,60])
print(d.count(20))
```

```python
from collections import deque
d = deque([10,20,30,40,50])
d.rotate(1)
print(d)
```

```python
from collections import deque
d = deque([11,22,33,44,55])
d.reverse()
print(d)
```
