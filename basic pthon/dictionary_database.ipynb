## Dictionary database 

```python
mydict = {
    "name" : "jay",
    "age" : 21,
    "city" : "surat"
}
print(mydict)
print(len(mydict))
print(type(mydict))
```

```python
mydict = {
    "name" : "jay",
    "age" : 21,
    "city" : "surat"
}
mydict["age"] = 23
print(mydict)
```

```python
mydict = {
    "name" : "jay",
    "age" : 21,
    "city" : "surat"
}
mydict.update({"age":25})
print(mydict)
```

```python
mydict = {
    "name" : "jay",
    "age" : 21,
    "city" : "surat"
}
mydict["color"] = "red"
print(mydict)
```

```python
mydict = {
    "name" : "jay",
    "age" : 21,
    "city" : "surat"
}
mydict.update({"color":"blue"})
print(mydict)
```

```python
mydict = {
    "name" : "jay",
    "age" : 21,
    "city" : "surat"
}
mydict.pop("age")
print(mydict)
```

```python
mydict = {
    "name" : "jay",
    "age" : 21,
    "city" : "surat"
}
mydict.popitem()
print(mydict)
```

```python
mydict = {
    "name" : "jay",
    "age" : 21,
    "city" : "surat"
}
del mydict
print(mydict)
```

```python
mydict = {
    "name" : "jay",
    "age" : 21,
    "city" : "surat"
}
mydict.clear()
print(mydict)
```

```python
mydict = {
    "name" : "jay",
    "age" : 21,
    "city" : "surat"
}
for i in mydict:
    print(i, mydict[i])
```

```python
mydict = {
    "name" : "jay",
    "age" : 21,
    "city" : "surat"
}
for i in mydict.keys():
    print(i)
```

```python
mydict = {
    "name" : "jay",
    "age" : 21,
    "city" : "surat"
}
for i in mydict.values():
    print(i)
```

```python
mydict = {
    "name" : "jay",
    "age" : 21,
    "city" : "surat"
}
for i in mydict.items():
    print(i)
```

```python
mydict = {
    "name" : "jay",
    "age" : 21,
    "city" : "surat"
}
new = dict(mydict)
print(new)
```

```python
mydict = {
    "name" : "jay",
    "age" : 21,
    "city" : "surat"
}
new = mydict.copy()
print(new)
```

```python
mydict = {
    "child1" : {
        "name" : "jay",
        "age" : 21,
    },
    "child2" : {
        "name" : "dip",
        "age" : 22,
    },
    "child3" : {
        "name" : "jaydip",
        "age" : 23
    }
}
print(mydict)
```

```python
child1 = {
    "name" : "jay",
    "age" : 21,
}
child2 = {
    "name" : "dip",
    "age" : 23,
}
child3 = {
    "name" : "jaydip",
    "age" : 23,
}
mydict = {
    "child1" : child1,
    "child2" : child2,
    "child3" : child3
}
print(mydict)
```

```python
mydict = {
    "child1" : {
        "name" : "jay",
        "age" : 21,
    },
    "child2" : {
        "name" : "dip",
        "age" : 22,
    },
    "child3" : {
        "name" : "jaydip",
        "age" : 23
    }
}
print(mydict["child2"]["name"])
```

```python
mydict = {
    "name" : "jay",
    "age" : 21,
    "city" : "surat"
}
new = mydict.get("city")
print(new)
```
