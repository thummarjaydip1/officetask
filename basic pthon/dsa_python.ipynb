#### Recursion

```python
def myfun(n):
    if n < 1:
        return n
    else:
        myfun(n-1)
        print(n)

myfun(10)
```

```python
def myfun(n):
    if n < 1:
        return n
    else:
        print(n)
        myfun(n-1)
myfun(10)
```

```python
def fact(n):
    if n<=1:
        return 1
    else:
        return n * fact(n-1)  # fact(5) ===> 5*4 => 5*4*3 => 5*4*3 => 5*4*3*2 => 5*4*3*2*1
    
print(fact(5))
print(fact(7))
print(fact(10))
```

```python
def max_number(number):
    if len(number) == 1:
        return number[0]

    max_num = max_number(number[1:])
    return number[0] if number[0] > max_num else max_num

a = [32,45,23,78,54,24,67,87,43]
print(max_number(a))
```
