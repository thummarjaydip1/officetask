## File Handling

```python
f = open("one.txt")
print(f)
```

```python
f = open("one.txt", "r")
f.close()
```

```python
f = open("one.txt", "r")
print("file name:", f.name)
print("file mode:", f.mode)
print("Is closed:", f.closed)

f.close()
print("Is closed:", f.closed)
```

```python
f = open("one.txt", "r")
a = f.read()
print(a)
```

```python
f = open("one.txt", "w")
f.write("hello python")
f.write(" file handling")
f.close()
```

```python
with open("one.txt", "r") as f:
    data = f.read()
    print(data)
```

```python
with open("one.txt", "a") as f:
    f.write("\nhello world")
print("file written successfully")
```
