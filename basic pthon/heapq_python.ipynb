## Heapq Python

```python
import heapq 
a = [25,35,10,30,20,15,40,50,45]
heapq.heapify(a)
print(a)
```

```python
import heapq
a = [25,35,10,30,20,15,40,50,45]
heapq.heappush(a, 5)
print(a)
```

```python
import heapq
a = [25,35,10,30,20,15,40,50,45]
heapq.heapify(a)
b = heapq.heappop(a)
print(b)
print(a)
```

```python
import heapq
a = [25,35,10,30,20,15,40,50,45]
heapq.heapify(a)

b = heapq.heappushpop(a, 55)
print(b)
print(a)
```

```python
import heapq
a = ["pineapple", "banana", "mango", "kiwi"]
heapq.heapify(a)                            # sorting a,b,c,d,e,f,...,z
heapq.heappush(a, "orange")
b = heapq.heappop(a)
print("removed:", b)
print(a)
```

```python
import heapq
a = [(2,"ram"), (1,"man"), (3, "jay")]
heapq.heapify(a)

b = heapq.heappushpop(a, (4, "yug"))
print(b)
print(a)
```

```python
import heapq 
a = [25,35,10,30,20,15,40,50,45]
b = heapq.nlargest(4, a)
print("4 largest element:", b)
```

```python
import heapq 
a = [25,35,10,30,20,15,40,50,45]
b = heapq.nsmallest(4, a)
print("4 smallest element:", b)
```

```python
import heapq
a = [25,35,10,30,20,15,40,50,45]
heapq.heapify(a)
b = heapq.heapreplace(a, 55)
print("remove:",b)
print(a)
```

```python
import heapq
a = [25,35,10,30,20,15,40,50,45]
b = [55,60,65,70]
c = tuple(heapq.merge(a, b))
print(c)

d = list(heapq.merge(sorted(a), sorted(b)))
print(d)
```
