## If elif else Match

```python
a = 5
if a > 0 :
    print("number is positive")
if a < 0:
    print("number is nagative")
if a==0:
    print("number is zero")
```

```python
num = 121
if num%2==0:
    print("even number")
else:
    print("odd number")
```

```python
age = int(input("enter your age: "))
if age >= 18:
    print("Eligible")
else:
    print("Not Eligible")
```

```python
a = int(input("enter any number"))
print(a)
if a > 0 :
    print("number is positive")
elif a < 0:
    print("number is nagative")
else:
    print("number is zero")
```

```python
a = 20
b = 33
if b > a:
  print("b is greater than a")
elif a == b:
  print("a and b are equal")
else:
  print("a is greater than b")
```

```python
day = 6
match day:
  case 1:
    print("Monday")
  case 2:
    print("Tuesday")
  case 3:
    print("Wednesday")
  case 4:
    print("Thursday")
  case 5:
    print("Friday")
  case 6:
    print("Saturday")
  case 7:
    print("Sunday")
```

```python
word = "start"
match word:
    case "stop":
        print("stoping")
    case "start":
        print("starting")
    case _:
        print("invalid command")
```
