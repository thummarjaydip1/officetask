```python
a = [2,4,6,8,10,12]
it = iter(a)
while True:
    try:
        no = next(it)
        print(no)
    except StopIteration:
        break
```

```python
class Myclass:
    def __iter__(self):
        self.a = 1
        return self
    def __next__(self):
        x = self.a
        self.a += 1
        return x
my = Myclass()
myit = iter(my)
print(next(myit))
print(next(myit))
print(next(myit))
print(next(myit))
```

```python
class Myclass:
    def __iter__(self):
        self.a = 1
        return self
    
    def __next__(self):
        if self.a <= 51:
            x = self.a
            self.a = self.a + 1
            if x%2 == 0:
                return x
    
        else:
            raise StopIteration

mycl = Myclass()
myit = iter(mycl)

for i in myit:
    if i == None:
        continue
    print(i, end=" ")
```

```python
class Myclass:
    def __iter__(self):
        self.a = 65
        return self
    
    def __next__(self):
        if self.a < 91:
            x = self.a
            self.a += 1
            return chr(x)
        else:
            raise StopIteration

mycl = Myclass()
myit = iter(mycl)

for i in myit:
    print(i, end="    ")
```
