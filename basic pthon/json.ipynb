## Python Json

```python
import json
x = '{"name":"jay","age":21,"city":"surat"}'
print(x)
print(type(x))

print()

y = json.loads(x)
print(type(y))
print(y)
print(y["name"])
print(y["age"])
print(y["city"])
```

```python
import json
a = {
    "name" : "jaydip",
    "age" : 21,
    "city" : "surat"
}
print(a)
print(type(a))

print()

b = json.dumps(a)
print(b)
print(type(b))
```

```python
import json
a = ["apple","banana","mango","kiwi"]
b = json.dumps(a)
print(a, type(a), b, type(b))

print()
a = ("apple","banana","mango")
b = json.dumps(a)
print(a, type(a), b, type(b))

print()
a = {
    "name" : "jay",
    "age" : 21,
    "city" : "surat"
}
b = json.dumps(a)
print(a, type(a), b, type(b))

print()
a = "hello world"
b = json.dumps(a)
print(a, type(a), b, type(b))

print()
a = 32
b = json.dumps(a)
print(a, type(a), b, type(b))

print()
a = 23.43
b = json.dumps(a)
print(a, type(a), b, type(b))

print()
a = True
b = json.dumps(a)
print(a, type(a), b, type(b))

print()
a = False
b = json.dumps(a)
print(a, type(a), b, type(b))

print()
a = None
b = json.dumps(a)
print(a, type(a), b, type(b))
```

```python
import json
x = {
    "name" : "jay",
    "age" : 21,
    "email" : "jay@gmail.com",
    "married" : True,
    "children" : ("mayur", "mayuri"),
    "address" : [
            {"city" : "bagasara"},
            {"pincode" : 365440}
        ]
}

y = json.dumps(x, indent=4)
print(type(y))
print(y)
```

```python
import json
x = {
    "name" : "jay",
    "age" : 21,
    "email" : "jay@gmail.com",
    "married" : True,
    "children" : ("mayur", "mayuri"),
    "address" : [
            {"city" : "bagasara"},
            {"pincode" : 365440}
        ]
}

y = json.dumps(x, indent=4, separators=("." , " = "))
print(type(y))
print(y)
```

```python
import json
x = {
    "name" : "jay",
    "age" : 21,
    "email" : "jay@gmail.com",
    "married" : True,
    "children" : ("mayur", "mayuri"),
    "address" : [
            {"city" : "bagasara"},
            {"pincode" : 365440}
        ]
}

y = json.dumps(x, indent=4, separators=("," , " = "), sort_keys=True)
print(type(y))
print(y)
```

```python
import json
data = {
    "name" : "jay",
    "age" : 21,
    "email" : "jay@gmail.com",
    "married" : True,
    "children" : ("mayur", "mayuri"),
    "address" : [
            {"city" : "bagasara"},
            {"pincode" : 365440}
        ]
}
with open("one.json", "w") as file:
    # json.dump(data, file, indent=4, sort_keys=True)
    json.dump(data, file, indent=4)

print("json file written successfully")
```
