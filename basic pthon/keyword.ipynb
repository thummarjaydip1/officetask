```python
# async, await
# import asyncio

async def greet():
  return "Hi"

async def main():
  msg = await greet()
  print(msg)

# asyncio.run(main())
main()
```

```python
# for, in, if, break, range
for i in range(8):
    print(i)
    if i == 5:
        break
```

```python
# continue 
for i in range(8):
    if i == 5:
        continue
    print(i)
```

```python
# match, case
command = "stop"
match command:
    case "start":
        print("starting")
    case "stop":
        print("stoping")
    case _:
        print("invalid command")
```

```python
# if, elif, else
a = int(input("enter any number:"))

if a%2==0:
    print(a, "is even numver")
elif a%2!=0:
    print(a, "is odd number")
else:
    print("please enter number")
```

```python
# try, except, else, finally
try:
    a = int(input("Enter any number"))
except:
    print("please enter int number")
else:
    print(a)
finally:
    print("this is finally block")
```

```python
# global
def func():
    global y
    y = "hello"

func()
print(y)
```

```python
# in
a = ["apple", "banana", "mango", "kiwi"]

print("mango" in a)
```

```python
# is
a = ["apple", "banana"]  # list
b = ("apple", "banana")  # tuple
print(a is b)
```

```python
# lambda
addition = lambda a, b: a + b
substraction = lambda a, b: a - b
multiplication = lambda a, b: a * b
division = lambda a, b: a / b
print(addition(100, 20))
print(substraction(100, 20))
print(multiplication(100, 20))
print(division(100, 20))
```
