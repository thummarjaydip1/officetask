## Loops

```python
fruits = ["apple", "banana", "cherry","kiwi"]
for x in fruits:
  print(x)
```

```python
for i in "banana":
    print(i)
```

```python
fruits = ["apple", "banana", "cherry","kiwi"]
for x in fruits:
  print(x)
  if x in "banana":
    break
```

```python
fruits = ["apple", "banana", "cherry","kiwi"]
for x in fruits:
  if x in "banana":
    continue
  print(x)
```

```python
for i in range(6):
    print(i)
```

```python
for x in range(2, 20, 2):
  print(x)
```

```python
for x in range(6):
  if x == 2: break
  print(x)
else:
  print("Finally finished!")
```

```python
color = ["red", "blue", "green"]
fruits = ["apple", "banana", "cherry"]

for x in color:
  for y in fruits:
    print(x, y)
```
