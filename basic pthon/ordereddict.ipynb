## OrderDict in Python

```python
from collections import OrderedDict

od = OrderedDict()
od["apple"] = 1
od["banana"] = 2
od["mango"] = 3

print(od)
print(od.items())
print(list(od.items()))
```

```python
from collections import OrderedDict

print("dict")
d = {}
d["a"] = 1
d["b"] = 2
d["c"] = 3
d["d"] = 4
for i, j in d.items():
    print(i, ":", j)

print("Ordered Dict")
od = OrderedDict()
od["d"] = 4
od["b"] = 2
od["a"] = 1
od["c"] = 3
for k, l in od.items():
    print(k, ":", l)
```

```python
from collections import OrderedDict

print("dict")
d = {"apple":1, "banana":2, "mango":3, "kiwi":4}
for i,j in d.items():
    print(i, ":", j)

print("OrderedDict")
od = OrderedDict([("kiwi",4),("banana",2),("apple",1),("mango",3)])
for i,j in od.items():
    print(i, ":", j)
```

```python
from collections import OrderedDict
od = OrderedDict([("apple",1),("banana",2),("mango",3),("kiwi",4)])
od["mango"] = 5
for i,j in od.items():
    print(i, ":", j)
```

```python
from collections import OrderedDict
od1 = OrderedDict([("apple",1),("banana",2),("mango",3)])
od2 = OrderedDict([("mango",3),("banana",2),("apple",1)])
if od1 == od2:
    print("both are equal")
else:
    print("both are not equal")
```

```python
from collections import OrderedDict
od1 = OrderedDict([("apple",1), ("banana",2),("mango",3)])
od2 = OrderedDict(reversed(list(od1.items())))
print(od2)
```

```python
from collections import OrderedDict
od = OrderedDict([("apple",1),("banana",2),("mango",3),("kiwi",4)])
print(od)

a = od.popitem(last=True)
print(a)
print(od)

b = od.popitem(last=False)
print(b)
print(od)
```

```python
from collections import OrderedDict
od = OrderedDict([("a",1),("b",2),("c",3),("d",4)])
print(od)

od.move_to_end("a")                 # Move to the last
od.move_to_end("c", last=False)     # Move to the front
print(od)
```

```python
from collections import OrderedDict
od = OrderedDict([("apple",1),("banana",2),("mango",3),("kiwi",4)])
od.pop("banana")
for k,v in od.items():
    print(k, v)

print("re-inserting ordereddict")
od["banana"] = 2
for k,v in od.items():
    print(k, v)
```
