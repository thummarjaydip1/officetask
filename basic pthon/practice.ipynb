```python
# intersection in set (all duplicate value print)
a = {"apple", "banana", "kiwi", "orange"}
b = {"lemon", "cherry", "kiwi", "orange"}
c = a.intersection(b)
print(c)
d = a & b
print(d)
```

```python
# symmetric difference (only print not same(common) element)
a = {"apple", "banana", "kiwi", "orange"}
b = {"lemon", "cherry", "kiwi", "orange"}
c = a.symmetric_difference(b)
print(c)
d = a ^ b
print(d)
```

```python
# decorator with argument
def entervalue(val):
    def calculate(fun):
        def inner(a, b):

            if val == "+":
                return fun(a + b)
            
            elif val == "-":
                return fun(a - b)

            elif val == "*":
                return fun(a * b)

            elif val == "/":
                return fun(a / b)        

        return inner
    return calculate

@entervalue("+")
def addition(x):
    return f"Addition is: {x}"
print(addition(20,10))

@entervalue("-")
def substraction(x):
    return f"substraction: {x}"
print(substraction(20,10))

@entervalue("*")
def multiplication(x):
    return f"multiplication: {x}"
print(multiplication(20,10))

@entervalue("/")
def division(x):
    return f"division: {x}"
print(division(20,10))
```
