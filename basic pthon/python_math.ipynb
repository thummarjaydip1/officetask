## Python math

```python
x = min(12,453,54,23,11,45,656)
print(x)

y = ("pineapple", "banana", "mango", "apple", "kiwi")
print(min(y))
```

```python
x = max(354,45,132,65,223,765,32,546,24)
print(x)

y = ("pineapple", "apple", "kiwi", "mango")
print(max(y))
```

```python
x = abs(-24.54)
y = abs(-543)
print(x)
print(y)
```

```python
a = pow(3,4)    # 3*3*3*3 = 81
b = pow(5,5)    # 5*5*5*5*5 = 3125
print(a)
print(b)
```

```python
import math

x = math.sqrt(81)
y = math.sqrt(2500)
print(x)
print(y)
```

```python
import math 
x = math.ceil(12.11)        # upward nearest val => 13
y = math.ceil(15.99)        # upward nearest val => 16
print(x)
print(y)
```

```python
import math
x = math.floor(23.25)       # downward nearest val => 23
y = math.floor(45.97)       # downward nearest val => 45
print(x)
print(y)
```
