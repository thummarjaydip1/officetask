##### 6. Write a Python program to calculate the factorial of a number using (a) a loop and (b) recursion.

###### a) loop

```python
n = int(input("enter int number:"))

fact = 1

for i in range(1, n+1):
    fact = fact * i
    
    
print("fact number:", fact)
```

###### b) recursion

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)

factorial(5)
```

##### 8. Write a Python program to display the Fibonacci series up to N terms using a function.

```python
try:
    def fibonacci(num):
        x = 0
        y = 1
        print(x, end=" ")
        for i in range(1, num+1):
            x, y = y, x+y
            print(x, end=" ")

    n = int(input("enter series number of fibonacci series:"))
    fibonacci(n)

except:
    print("please enter integer value")
```

##### 9. Write a Python program that converts temperatures between Celsius, Fahrenheit, and Kelvin using separate functions.

```python
def temprature(celsius):
    print("celsius is:", celsius)

    fehrenheit = (celsius * 9/5) + 32
    print(f"celsius convert to fehrenheit is: {fehrenheit:.2f}")

    kelvin = celsius + 273.15
    print("celsius convert to kelvin is:", kelvin)

temp = int(input("enter temperature in celsius:"))
temprature(temp)
```

##### 11. Write a Python program to manage student records using a list. The program should allow adding, displaying, searching, and deleting student records

```python
l1 = []

def add():
    b = input("enter add student name:")
    print("added student is:", b)
    l1.append(b)

def display():
    print('student list is:', l1)

def search():
    c = input("enter search student:")
    print("search student is:", c)
    if c in l1:
        d = l1.index(c)
        print("search student index number is:", d)
    else:
        print("student not found")

def delete():
    e = input("enter delete student:")
    print("deleted student is:", e)
    if e in l1:
        l1.remove(e)
        print(l1)
    else:
        print("student not found")

while True:
    a = input("enter ('add','display','search','delete')")
    if a == "add":
        add()
    elif a == "display":
        display()
    elif a == "search":
        search()
    elif a == "delete":
        delete()
    elif a == "exit":
        break
    else:
        print("invalid choice please enter ('add', 'display', 'search', 'delete') ")
```
