## Exception Handling in Python

```python
try:
    a = 15 / 0
    print(a)
except ZeroDivisionError:
    print("divide by zero generate exception")
```

```python
try:
    a = int(input("enter first value:"))
    b = int(input("enter second value:"))
    print("first value:", a)
    print("second value:", b)
    ans = str(a) / b
except ValueError:
    print("value error generated")
except TypeError:
    print("type error generated")
except ZeroDivisionError:
    print("divide by zero exception generate")
else:
    print("division is:", ans)
finally:
    print("this is finally block")
```

```python
try:
    x = int("string")
    ans = 15 / x
except ValueError:
    print("Not valid generated value error")
except ZeroDivisionError:
    print("zero divide by error")
```

```python
list1 = ["10", "hello", True, 12.32]
try:
    total = int(list1[0]) + int(list1[1])
    print(total)
except (ValueError, TypeError) as e:
    print("Error", e)

except IndexError:
    print("Index error generated")
```

```python
try:
    a = "hello" / 23
    print(a)
except ArithmeticError:
    print("Arithmetic error")
except:
    print("somthing went wrong")
```

```python
def myfun(a):
    try:
        x = int(a)
    except ValueError:
        print("value error exception generated")
    except:
        print("exception generated")

    else:
        if x%2==0:
            print("entered number is even number:", x)
        else:
            print("entered number is odd number:", x)

a = input("enter first value:")
myfun(a)
```

```python
x = -1
if x < 0:
    raise Exception("number is nagative")
```

```python
x = int(input("enter any no"))
y = int(input("enter any no"))
ans = x / y                          # ans datattype is float
print(ans)
if type(ans) is not int:
    raise Exception("answer is not int number")
```

```python
# BaseException
try:
    raise BaseException("This is base exception")
except BaseException as e:
    print(e)
```

```python
# Exception
try:
    raise Exception("This is generic exception")
except Exception as e:
    print(e)
```

```python
# ArithmeticError
try:
    raise ArithmeticError("Arithmetic error occurred")
except ArithmeticError as e:
    print(e)
```

```python
# ZeroDivisionError
try:
    res = 24 / 0
except ZeroDivisionError as e:
    print(e)
```

```python
# OverflowError
import math
try:
    result = math.exp(7000)  
    print(result)
except OverflowError as e:
    print(e)
```

```python
# FloatingPointError
import numpy as np
np.seterr(all="raise")

try:
    np.divide(1,0)
except FloatingPointError as e:
    print("floating error:", e)
```

```python
# AssertionError
try:
    assert 2!=2, "Assertion fail"
    # assert 3==2, "Assertion fail"
except AssertionError as e:
    print(e)
```

```python
# AttributeError
class Myclass:
    def one():
        return "hello"
    
obj = Myclass()

try:
    obj.disp()              # obj.one() exception not generated
except AttributeError as e:
    print(e)
```

```python
# IndexError
l1 = [23,435,56,87]
try:
    print(l1[5])
except IndexError as e:
    print(e)
```

```python
# KeyError
dic = {
    "name" : "jay",
    "age" : 21,
    "city" : "surat"
}
try:
    val=dic["marks"]
except KeyError as e:
    print(e)
```

```python
# NameError
try:
    print(variable)
except NameError as e:
    print(e)
```

```python
# OSError, FileNotFoundError
try:
    open("onefile.json")
# except FileNotFoundError as e:
#     print("file not found error:", e)
except OSError as e:
    print("OSError caught:", e)
```

```python
# TypeError
try:
    a = "jay"
    b = 4
    ans = a + b
except TypeError as e:
    print(e)
```

```python
# ValueError
try: 
    a = "jay" 
    int(a)
except ValueError as e:
    print(e)
```

```python
try:
    a = 5
    print("positive" a)
except SyntaxError as e:
    print(e)
```

```python
# ImportError
try:
    from python_module import new_module
except ImportError as e:
    print(e)
```

```python
# ModuleNotFoundError
try:
    import new
except ModuleNotFoundError as e:
    print(e)
```
