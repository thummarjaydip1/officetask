## Variables

```python
a = "jay"
b = 12
print(a)
print(b)
```

```python
a = 243
a = "raj"
print(a)
```

```python
a = str(5)
b = int(5)
c = float(5)
print(a)
print(b)
print(c)
```

```python
x = 5
y = "5"
z = 5.0
print(type(x), x)
print(type(y), y) 
print(type(z), z) 
```

```python
age = 21
_colour = "red"
total_score = 75
print(age)
print(_colour)
print(total_score)
```

```python
a = b = c = 10000
print(a,b,c)
```

```python
a, b, c = "hello", 12, 13.5
print("a =", a)
print("b =", b)
print("c =", c)
```

```python
a = 1
b = a
b = b + 1
print(a)
print(b)
```

```python
a, b = 10, 20
print("a:", a)
print("b:", b)
a, b = b, a
print("a:", a)
print("b:", b)
```
