### Python File Handling

###### a mode not overhide text write text or created file does not exists
###### w mode overhide text write text or created file does not exists
###### x mode are create file
###### r mode are read file
###### write() method write file
###### read() method read file
###### open() method open file
###### close() method clos file

```python
with open('demofile.txt','a') as file:
    file.write("\nhello world!")
print("writed file")
```

```python
with open("demofile.txt","w") as file:
    file.write("\nhow are you?")
print("writed file")
```

```python
# file= open("myfile.txt","x")
file = open("myfile.txt","x")
print(f"{file} is created")
```

```python
# file = open("C:\\office\file handling\demofile.txt")
file = open("demofile.txt") # default read mode j hoy che 'r'
print(file.read())
```

```python
with open("demofile.txt","r") as file:
    print(file.read(7))
```

```python
with open("demofile.txt") as f:
  print(f.readline()) # first line printed
  print(f.readline()) # second line printed
f.close()
```

```python
file = open("demofile.txt","r")
for x in file:
    print(x)
file.close()
```
