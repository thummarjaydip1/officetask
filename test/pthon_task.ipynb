##### 4. Write a Python function to reverse a list in place.

```python
def myfun(list1):
    # list1.reverse()
    list1.sort(reverse=True)
    print(list1)

li = [11,22,33,44,55]
myfun(li)
```

##### 5. Write a Python function to find the sum of all elements in a list.

```python
def myfun(list1):
    total = 0
    for i in list1:
        total += i
    print(total)

li = [11,22,33,44,55]
myfun(li)
```

#### Dictionary

##### 1. Write a Python function to add a key-value pair to a dictionary.

```python
def myfun(dictionary, key, value):
    dictionary[key] = value
    print(dictionary)

mydict = {
    "name" : "jay",
    "age" : 21,
    "email" : "jay@gamil.com"
}
key = "city"
value = "surat"
myfun(mydict, key, value)
```

##### 3. Write a Python function to check if a key is present in a dictionary.

```python
def myfun(dictionary, key):
    if key in dictionary:
        print("Key is present in dictionary", key, ":", dictionary[key])
    else:
        print("Key is not present in dictionary", key)

mydict = {
    "name" : "jay",
    "age" : 21,
    "email" : "jay@gamil.com"
}
key = "age"
myfun(mydict, key)
```

##### 4. Write a Python function to get the value associated with a specified key from a dictionary.

##### 5. Write a Python function to get a list of all the keys in a dictionary.

```python
li = []
def myfun(dictionary):
    for i in dictionary.keys():
        li.append(i)
    print(li)

mydict = {
    "name" : "jay",
    "age" : 21,
    "email" : "jay@gmail.com",
    "city" : "surat"
}
myfun(mydict)
```

#### Sets

##### 1. Write a Python function to add an element to a set.

```python
def myfun(myset, element):
    myset.add(element)
    print(myset)

set1 = {"apple", "banana", "mango", "orange"}
ele = "kiwi"
myfun(set1, ele)
```

##### 2. Write a Python function to remove an element from a set.

```python
def myfun(myset, element):
    myset.remove(element)
    # myset.discard(element)
    print(myset)

set1 = {"apple", "banana", "mango", "orange"}
ele = "banana"
myfun(set1, ele)
```

##### 3. Write a Python function to check if an element is present in a set.

```python
def myfun(myset, element):
    if element in myset:
        print("element is present in set element is:", element)
    else:
        print("element is not present in set element is:", element)

set1 = {"apple", "banana", "mango", "orange"}
ele = "banana"
myfun(set1, ele)
```

##### 4. Write a Python function to find the intersection of two sets.

```python
def myfun(set1, set2):
    set3 = set1.intersection(set2)
    print(set3)

set1 = {"apple", "banana", "kiwi", "orange"}
set2 = {"watermelon", "kiwi", "banana", "mango"}
myfun(set1, set2)
```

##### 5. Write a Python function to find the union of two sets.

```python
def myfun(set1, set2):
    set3 = set1.union(set2)
    print(set3)

set1 = {"apple", "banana", "kiwi", "orange"}
set2 = {"watermelon", "kiwi", "banana", "mango"}
myfun(set1, set2)
```

#### String

##### 1. Write a Python function to find the length of a string.

```python
def myfun(string):
    a = len(string)
    print(a)

string = "Hello World"
myfun(string)
```

##### 2. Write a Python function to concatenate two strings.

```python
def myfun(string1, string2):
    a = string1 + string2
    print(a)

string1 = "Hello user,"
string2 = " How are you?"
myfun(string1, string2)
```

##### 3. Write a Python function to find the index of the first occurrence of a character in a string.

```python
def myfun(string, chr):
    for i in range(len(string)):
        if string[i] == chr:
            return i
    return -1

string = "Hello World"
chr = 'o'
myfun(string, chr)
```

##### 4. Write a Python function to convert a string to lowercase.

```python
def myfun(string, chr):
    for i in range(len(string)):
        if string[i] == chr:
            return i
    return -1

myfun(string, chr)
```

##### 5. Write a Python function to check if a string is a palindrome.


```python
def myfun(string):
    a = string[0:]
    b = string[::-1]
    if a == b:
        print("String is Palindrome")
    else:
        print("String is not Palindrome")

string = "aabbaa"
myfun(string)
```

#### Loops and if-else statements

##### 1. Write a Python function to print all even numbers less than or equal to a specified limit using a for loop and an if statement.

```python
def myfun(list1, limit):
    for i in list1:
        if i <= limit:
            if i%2==0:
                print(i)

li = [12,35,54,32,65,30,74,23,24,86,18]
limit = 50
myfun(li, limit)
```

##### 2. Write a Python function to print the factorial of a given number using a for loop.

```python
def myfun(num):
    fact = 1
    for i in range(1, num+1):
        fact *= i
    print(fact)

myfun(3)
myfun(5)
myfun(7)
```

##### 3. Write a Python function to find the largest element in a list of integers using a for loop and an if statement.

```python
def myfun(list1):
    if len(list1) == 1:
        return list1[0]
    max_num = 0
    for i in list1:
        if max_num <= i:
            max_num = i
    return max_num

li = [32,54,12,54,76,87,42,76,98,43,53]
print(myfun(li))
```

##### 4. Write a Python function to check if a number is prime using a for loop and an if-else statement.

```python
def myfun(number):
    if number>0:
        if number == 1 or number == 2:
            print("not prime number")

        for i in range(3,number):
            if number%i == 0:
                print("not prime")
                break
            else:
                print("prime number")
                break
        
    else:
        print('please enter positive integer number')
        
myfun(2)
```

##### 5. Write a Python function to check if a list of integers is sorted in non-decreasing order using a for loop and an if statement.

```python
def myfun(list1):
    for i in range(len(list1)):
        if list1[i] > list1[i+1]:
            return "non deceasing order list"

    return "decreasing list"
    
li = [24,53,23,65,23,87,12,54,78,32]
myfun(li)
```
